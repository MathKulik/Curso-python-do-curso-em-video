```python
p=float(input('me fala o valor do produto '))
f=str(input('me fala a forma de pagamento')).lower()
if f=='vista' or f=='dinheiro' or f=='cheque' or f=='pix':
    v=p*0.1
    vf=p-v
    print(f'o valor final fica {vf}')
elif f=='debito':
    v=p*0.05
    vf=p-v
    print(f'o valor final fica {vf}')
elif f=='parcela 3x':
    v=p*0.2
    vf=p+v
    print(f'o valor final fica {vf}')
else:
    print('nenhum desconto ou juros foi aplicado')
```
