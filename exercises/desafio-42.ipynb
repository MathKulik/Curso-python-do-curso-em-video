```python
a=float(input('me fala o valor do primeiro lado'))
b=float(input('me fala o valor do segundo lado'))
c=float(input('me fala o valor do terceiro lado'))

if a+b>c and a+c>b and b+c>a:
    print('é um triangulo')
    if a==b==c:
        print('é um triangulo equilatero')
    elif a!=b!=c:
        print('é um triangulo isósceles')
    else:
        print('é um triangulo escaleno')
else:
    print('não é um triangulo')
```
