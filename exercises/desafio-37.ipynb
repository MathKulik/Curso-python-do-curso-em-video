```python
n=int(input('me fala o numero para converter '))
con=int(input('escolha de 1 a 3 para selecionar as opções: Binário, Octal e Hexadecimal '))
if con==1:
    bi=bin(n)
    print(f'o numero é {n} e o numero convertido é {bi}')
elif con==2:
    oc=oct(n)
    print(f'o numero é {n} e o numero convertido é {oc}')
elif con==3:
    he=hex(n)
    print(f'o numero é {n} e o numero convertido é {he}')
else:
    print('invalido')
    
```
