```python
#exercicio 017

from math import sqrt
a = float(input('me fala o cateto oposto'))
b = float(input('me fala o cateto adjacente'))
h = sqrt(a**2+b**2) #existe um modulo exclusivo para hipotenusa
print('o seu oposto é {} e o seu adjacente é {} e isso da {}'.format(a,b,h))
```

```python
#exercicio 018

from math import sin, cos, tan, radians
a = int(input('me fala o angulo'))
sn = sin(radians(a))
cs = cos(radians(a))
tn = tan(radians(a))
print('o se angulo é {} logo o seu seno é {:.2f}, seu cosseno é {:.2f} e a sua tangente é {:.2f}'.format(a, sn, cs,tn))
```
